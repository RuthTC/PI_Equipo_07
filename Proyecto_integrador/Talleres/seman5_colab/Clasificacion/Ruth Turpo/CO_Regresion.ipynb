```python
# 1. Definir la variable objetivo (por ejemplo, Daily AQI Value)
Y = df['Daily AQI Value']

# 2. Definir 'x' seleccionando solo columnas de tipo número (excluyendo la variable objetivo)
x = df.select_dtypes(include=['number']).drop(columns=['Daily AQI Value'])

# 3. Limpiar posibles valores vacíos (NaN) con 0
x = x.fillna(0)

x.head()
```

```python
Y.head()
```

```python
from sklearn.model_selection import train_test_split
# divide datos en: entrenamiento y prueba
```

```python
x_train, x_test, Y_train, Y_test = train_test_split(
    x,Y, test_size=0.3, random_state=123
)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn import metrics
```

```python
lm = LinearRegression()
```

```python
lm.fit(x_train, Y_train)
```

```python
print(lm.intercept_)
```

```python

# Verificamos el modelo lineal
print(lm.coef_)
```

```python
cdf=pd.DataFrame(data=lm.coef_, index= x_train.columns, columns=["Coefficients"])
cdf
```

```python
n = x_train.shape[0]
k = x_train.shape[1]
dfN = n - k

train_pred = lm.predict(x_train)
train_error = np.square(train_pred - Y_train)
sum_error = np.sum(train_error)

# Inicializar lista con la longitud dinámica k
se = []

for col in x_train.columns:
    r = sum_error / dfN
    r = r / np.sum(np.square(x_train[col] - x_train[col].mean()))
    se.append(np.sqrt(r))

cdf["Standard Error"] = se
cdf["t-statistic"] = cdf["Coefficients"] / cdf["Standard Error"]
cdf
```

```python
l = list(cdf.index)

from matplotlib import gridspec
fig = plt.figure(figsize=(18, 10))
gs = gridspec.GridSpec(2, 2)

ax0 = plt.subplot(gs[0])
ax0.scatter(df[l[0]], df['Daily Max CO Concentration'])
ax0.set_title(l[0] + " vs. Daily Max CO Concentration", fontdict={'fontsize':20})

ax1 = plt.subplot(gs[1])
ax1.scatter(df[l[1]], df['Daily Max CO Concentration'])
ax1.set_title(l[1] + " vs. Daily Max CO Concentration", fontdict={'fontsize':20})

ax2 = plt.subplot(gs[2])
ax2.scatter(df[l[2]], df['Daily Max CO Concentration'])
ax2.set_title(l[2] + " vs. Daily Max CO Concentration", fontdict={'fontsize':20})

ax3 = plt.subplot(gs[3])
ax3.scatter(df[l[3]], df['Daily Max CO Concentration'])
ax3.set_title(l[3] + " vs. Daily Max CO Concentration", fontdict={'fontsize':20})
```

```python
from ast import increment_lineno
import matplotlib.pyplot as plt
%matplotlib inline
import pandas as pd
import numpy as np

# 1. Cargar tu archivo CSV real de Calidad del Aire
df = pd.read_csv('CO_daily_aqs_data_downloaded_2026-09-17 19_18_34.csv')

# 2. Filtrar únicamente las columnas numéricas de tu dataset
df = df.select_dtypes(include=['number']).dropna(axis=1, how='all').fillna(0)

# Mostrar los primeros 10 registros reales
df.head(10)
```

```python
# Asignar la última columna numérica como variable a predecir (y)
target_col = df.columns[-1]

y = df[target_col]
x = df.drop(columns=[target_col])

n_features = x.shape[1]

print(f"Variable a predecir (y): {target_col}")
print(f"Número de características reales en X: {n_features}")
print("Columnas de X:", list(x.columns))
```

Organización de los datos y elaboración de visualizaciones

```python
with plt.style.context('Solarize_Light2'):
    for i, col in enumerate(x.columns):
        plt.figure(figsize=(6, 4))
        plt.grid(True)
        plt.xlabel('Característica: ' + col, fontsize=10)
        plt.ylabel('Salida real: ' + y.name, fontsize=10)
        plt.scatter(df[col], y, c='red', s=50, alpha=0.6)
        plt.title(f'{col} vs {y.name}', fontsize=12)
        plt.show()
```

Árbol de decisión para regresión

Entrenamiento del Árbol y Primer Gráfico (Real vs Predicho)


```python
from sklearn import tree
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

# Dividir los datos
X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=123
)

# Creación del modelo
tree_model = tree.DecisionTreeRegressor(
    max_depth=5,
    random_state=10
)

# Entrenar modelo
tree_model.fit(X_train, y_train)

# Hacemos predicciones
test_pred = tree_model.predict(X_test)

# Real vs predicho
plt.figure(figsize=(6, 4))
plt.scatter(x=y_test, y=test_pred)

plt.xlabel(f"Valor real ({y.name})")
plt.ylabel("Valor predicho")
plt.title(f"Real vs Predicho: {y.name}")
plt.grid(True)
plt.show()

# Error cuadrático medio
print(
    "Mean square error (MSE):",
    metrics.mean_squared_error(y_test, test_pred)
)
```

Cálculo del Error Cuadrático Medio (MSE) y Explicación

```python
from sklearn import metrics

# Hacemos predicciones
test_pred = tree_model.predict(X_test)

plt.scatter(x=y_test, y=test_pred)  # real vs predicho
plt.xlabel(f"Valor real ({y.name})")
plt.ylabel("Valor predicho")
plt.title(f"Real vs Predicho: {y.name}")
plt.grid(True)
plt.show()

print(
    "Mean square error (MSE):",
    metrics.mean_squared_error(y_test, test_pred)
)

# Error cuadrático medio
'''
¿Por qué se eleva al cuadrado?

Para que los errores positivos y negativos no se cancelen
y para penalizar más los errores grandes.
'''
```

Importancia Relativa de las Características


```python
import numpy as np
import matplotlib.pyplot as plt

n_features = x.shape[1]

print("Importancia relativa de las características: ", tree_model.feature_importances_)

with plt.style.context('dark_background'):
    plt.figure(figsize=(10, 7))
    plt.grid(True)
    plt.yticks(range(n_features, 0, -1), x.columns, fontsize=12)
    plt.xlabel("Importancia relativa (normalizada) de los parámetros", fontsize=15)
    plt.ylabel("Características\n", fontsize=15)
    plt.barh(range(n_features, 0, -1), width=tree_model.feature_importances_, height=0.5)
    plt.tight_layout()
    plt.show()
```
